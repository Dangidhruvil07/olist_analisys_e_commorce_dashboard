```python
# 1. customers → orders
unmatched_customer_ids = orders[
    ~orders['customer_id'].isin(customers['customer_id'])
]

print("customers → orders:", len(unmatched_customer_ids))


# 2. orders → order_items
unmatched_order_items = order_items[
    ~order_items['order_id'].isin(orders['order_id'])
]

print("orders → order_items:", len(unmatched_order_items))


# 3. orders → order_payments
unmatched_payments = order_payment[
    ~order_payment['order_id'].isin(orders['order_id'])
]

print("orders → order_payments:", len(unmatched_payments))


# 4. orders → order_reviews
unmatched_reviews = order_review[
    ~order_review['order_id'].isin(orders['order_id'])
]

print("orders → order_reviews:", len(unmatched_reviews))


# 5. products → order_items
unmatched_products = order_items[
    ~order_items['product_id'].isin(products['product_id'])
]

print("products → order_items:", len(unmatched_products))


# 6. sellers → order_items
unmatched_sellers = order_items[
    ~order_items['seller_id'].isin(sellers['seller_id'])
]

print("sellers → order_items:", len(unmatched_sellers))
```
