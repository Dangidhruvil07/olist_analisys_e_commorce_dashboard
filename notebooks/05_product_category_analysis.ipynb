```python
df= order_product_details.groupby('product_category_name_english')['price'].sum().reset_index().sort_values(by='price', ascending=False)
```

```python
op_df=df.merge(temp_df, on='product_category_name_english', how='left')
```

```python
op_df['Avg_sels_values_per_order']=op_df['price']/op_df['order_id']
```

```python
avg_sales = op_df['price'].mean()
avg_orders = op_df['order_id'].mean()
print(avg_sales)
print(avg_orders)
```

```python
def category_classfication(row):
    if row['price'] > avg_sales and row['order_id'] > avg_orders:
        return 'High + High'
    elif row['price'] < avg_sales and row['order_id'] < avg_orders:
        return 'Low + Low'
    elif row['price'] > avg_sales and row['order_id'] < avg_orders:
        return 'High + Low'
    else:
        return 'Low + High'
```

```python
op_df['segment'] = op_df.apply(category_classfication,axis=1)
op_df['segment'].value_counts()
```
